# PREPROCESSING

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_selector, make_column_transformer
from sklearn.impute import SimpleImputer
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df.head()
```

```python
numerical_pipeline = make_pipeline(SimpleImputer(strategy='mean'), StandardScaler())
```

```python
preprocessor = make_column_transformer((numerical_pipeline, num_var))
```

```python
df_scaled = preprocessor.fit_transform(df)
```

```python
print(type(df_scaled))  # numpy.ndarray, pas un DataFrame
print(df_scaled) 
```

```python
colonnes = preprocessor.get_feature_names_out()
```

```python
print(colonnes)
```

```python
df_scaled = pd.DataFrame(df_scaled, columns=colonnes)
```

```python
df_scaled.head()
```

```python
# Détermination du nombre optimal de clusters (méthode du coude + silhouette)
inertias = []
silhouette_scores = []
k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df_scaled)
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(df_scaled, kmeans.labels_))

# Visualisation pour choisir le nombre de clusters
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

ax1.plot(k_range, inertias, 'bo-')
ax1.set_xlabel('Nombre de clusters (k)')
ax1.set_ylabel('Inertie')
ax1.set_title('Méthode du coude')
ax1.grid(True, alpha=0.3)

ax2.plot(k_range, silhouette_scores, 'ro-')
ax2.set_xlabel('Nombre de clusters (k)')
ax2.set_ylabel('Score de silhouette')
ax2.set_title('Score de silhouette')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Sélection du nombre optimal de clusters
optimal_k = k_range[np.argmax(silhouette_scores)]
print(f"\nNombre optimal de clusters: {optimal_k}")
print(f"Score de silhouette: {max(silhouette_scores):.3f}")

# Clustering final
final_kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
cluster_labels = final_kmeans.fit_predict(df_scaled)

# Ajout des labels de cluster au DataFrame original (non transformé)
df['cluster'] = cluster_labels

## 6. ANALYSE DES SEGMENTS
print(f"\n=== ANALYSE DES SEGMENTS ===")
segment_analysis = df.groupby('cluster')[num_var].agg(['mean', 'std']).round(2)
print("Statistiques par segment:")
print(segment_analysis)

# Pourcentage de clients par segment
segment_sizes = df['cluster'].value_counts().sort_index()
segment_percentages = (segment_sizes / len(df) * 100).round(1)
print(f"\nRépartition des clients par segment:")
for cluster, percentage in segment_percentages.items():
    print(f"Segment {cluster}: {segment_sizes[cluster]} clients ({percentage}%)")
```

Logique métier appliquée aux clusters
En observant vos données :

Cluster 0 (54.7%) : Revenu moyen plus faible (~36.8k), 1.26 enfant en moyenne, faible dépense (Monetary ~144) et faible fréquence d'achat (Frequence ~9).

Cluster 1 (45.3%) : Revenu moyen élevé (~70.7k), peu d'enfants (0.58), très fortes dépenses (Monetary ~1162) et fréquence d'achat élevée (Frequence ~21).

```python

# ==========================================
# FONCTION 1 : Segmentation Principale (Cluster -> Persona Macro)
# ==========================================

#Au vu des statistiques moyennes pour chaque cluster , on peut définir le cluster 1 comme étant le cluster pour une clientèle Econome , et pour la sécond 
#Classe comme unc clientèle assez aisée et depensif comme notre premium.
def assigner_personas_base(df: pd.DataFrame) -> pd.DataFrame:
    """
    Assigne le persona macro de base basé strictement sur les 2 clusters K-Means.
    """
    df_res = df.copy()
    
    macro_map = {
        0: "Macro-Persona: Clientèle Économe",
        1: "Macro-Persona: Clientèle Premium VIP"
    }
    
    df_res['macro_persona'] = df_res['cluster'].map(macro_map)
    return df_res


# ==========================================
# FONCTION 2 : Segmentation Avancée & Stratégies Marketing (Sous-segmentation)
# ==========================================
def affiner_segmentation_et_strategie(df: pd.DataFrame) -> pd.DataFrame:
    """
    Complète la première fonction en découpant les macro-personas en sous-groupes ultra-ciblés
    et en associant la stratégie marketing et les canaux de communication adéquats.
    """
    # Vérification que la fonction 1 a bien été exécutée
    if 'macro_persona' not in df.columns:
        df = assigner_personas_base(df)
        
    df_res = df.copy()
    
    # Calcul des médianes internes pour créer des seuils dynamiques et précis
    mediane_recence_c1 = df_res[df_res['cluster'] == 1]['Recency'].median()
    mediane_recence_c0 = df_res[df_res['cluster'] == 0]['Recency'].median()

    # 1. Définition des conditions de sous-segmentation
    conditions = [
        # Sous-segments Cluster 1 (Premium / Haut Revenu)
        (df_res['cluster'] == 1) & (df_res['Recency'] <= mediane_recence_c1),
        (df_res['cluster'] == 1) & (df_res['Recency'] > mediane_recence_c1),
        
        # Sous-segments Cluster 0 (Économe / Revenu Modéré)
        (df_res['cluster'] == 0) & (df_res['Number_kids'] > 0) & (df_res['Recency'] <= mediane_recence_c0),
        (df_res['cluster'] == 0) & (df_res['Number_kids'] > 0) & (df_res['Recency'] > mediane_recence_c0),
        (df_res['cluster'] == 0) & (df_res['Number_kids'] == 0)
    ]

    # 2. Libellés des Personas Détaillés
    personas_detailles = [
        "VIP Actif & Loyal",
        "VIP en Risque de Churn",
        "Famille Économe Active",
        "Famille Économe Dormante",
        "Acheteur Individuel Occasionnel"
    ]

    # 3. Stratégies Marketing Ciblées
    strategies = [
        "Programme Ambassadeur : Ventes privées exclusives, cadeaux VIP et accès anticipé aux collections.",
        "Campagne de Réengagement VIP : Offres personnalisées haut de gamme et service client dédié pour relancer l'intérêt.",
        "Promotions Volume/Famille : Bons de réduction sur les achats groupés, offres sur les gammes enfants/foyer.",
        "Relance Promo Impact : Codes promo à durée limitée par SMS, ventes flash à petits prix.",
        "Cross-Selling d'Entrée de Gamme : Recommandations basées sur le panier moyen bas, incitation au 2ème achat."
    ]

    # 4. Canaux Recommandés
    canaux = [
        "Email VIP personnalisé, Appel direct / Concierge",
        "Email dédié personnalisé, Direct Mail (Courrier physique)",
        "Newsletter Email hebdomadaire, Application Mobile",
        "SMS Marketing, WhatsApp Business",
        "Retargeting Publicitaire (Social Ads), Notifications Push"
    ]

    # Application des règles
    df_res['persona_detaille'] = np.select(conditions, personas_detailles, default="Non Classé")
    df_res['strategie_marketing'] = np.select(conditions, strategies, default="Stratégie Générique")
    df_res['canal_recommande'] = np.select(conditions, canaux, default="Email Standard")

    return df_res
```

```python
# 1. Exécution de la première fonction
df_macro = assigner_personas_base(df)

# 2. Exécution de la deuxième fonction pour enrichir la première
df_final = affiner_segmentation_et_strategie(df_macro)

# Visualisation du résultat
colonnes_a_afficher = ['cluster', 'macro_persona', 'persona_detaille', 'strategie_marketing', 'canal_recommande']
print(df_final[colonnes_a_afficher].head())
```
